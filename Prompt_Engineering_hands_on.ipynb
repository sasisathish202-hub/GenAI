# ReAct PROMPTING - the prompt encourage the model to reason through a problem step-by-step before giving the final answer

```python
from openai import OpenAI

openai_api_key = os.getenv("Your API KEY")
MODEL = "gpt-5-nano"

question = input("Enter User Question:")

prompt = f"""
You are an Technical assistant.

Solve the user's problem using this process:

1. Understand the problem.
2. Identify the important information.
3. Determine the best approach.
4. Check the result.
5. Give the final answer.

Do not provide hidden chain-of-thought.
Instead, provide only a short reasoning summary followed by
the final answer.

User Question:
{question}

Format:

Reasoning Summary:
<short explanation>

Final Answer:
<answer>
"""

response = client.responses.create(
    model = MODEL,
    input = prompt
)

print("\n Response:")
print(response.output_text)
```

# PROMPT CHAINING - Use the output of one LLM prompt as the input to input prompt:

```python
from openai import OpenAI

openai_api_key = os.getenv("Your API KEY")
MODEL = "gpt-5-nano"

question = input("Enter User Question: ")


# CHAIN-1 - Generate Explanation

prompt_1 = f"""
Explain the following user question.

Give the key concepts and important points.
Do not write the final polished answer yet.

User Question:
{question}
"""
response_1 = client.responses.create (
    model = MODEL,
    input = prompt_1
)
intermediate_result = response_1.output_text

# CHAIN -2 - Create Final Answer

prompt_2 = f"""
Create a clear and simple final answer using the
information generated by the first prompt.

Original User Question:
{question}

Information from Chain 1:
{intermediate_result}

-Give only the final answer.
-Keep it short and simple.
-Maximum 5 sentences.
"""

response_2 = client.responses.create(
    model = MODEL,
    input = prompt_2
)

final_answer = response_2.output_text

print("\n Response:")
print(final_answer)
```

# DYNAMIC TEMPLATE- Dynamically changes the prompt based on user input:

```python
from openai import OpenAI

openai_api_key = os.getenv("Your API KEY")
MODEL = "gpt-5-nano"

question = input ("Enter User Question")

style = input ("Enter Response Style " 
               "(summarise / translate / explain / code):"
              )

templates = {
    "summarize" : "Summarize the following text in 3 concise bullet point.",
    "translate" : "Translate the following text into French.",
    "explain" : " Explain the following concept clearly.",
    "code" : "Write only a short python code for the following task. Do not explain the code"
}

selected_template = templates.get(style.lower(), 
                                 templates["explain"]
                                )
prompt = f"""
You are a helpful AI assistant.

{selected_template}

User Question:
{question}
"""

response = client.responses.create(
    model = MODEL,
    input = prompt
)
print("\n Response:")
print(response.output_text)
```
