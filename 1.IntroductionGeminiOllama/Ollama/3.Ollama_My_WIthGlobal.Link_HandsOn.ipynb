```python
import ollama 
import gradio as gr

system_msg = "Yor are a financial expert specializing in real estate investment and negotiation."

def CustomLlama3(user_input):

    messages = [
        { "role": "system", "content": system_msg},
        {"role": "user", "content": user_input}
    ]

    response = ollama.chat(
        model = "llama3.2", 
        messages = messages)

    return response["message"]["content"]

demo = gr.Interface(fn = CustomLlama3, inputs = "text", outputs = "text",
                    title = "My Saving Assistant")

demo.launch(share = True)
                                            
```
